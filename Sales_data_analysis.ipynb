Coversion of Date columns from **Object** to ***datetime** format for the calculation of 
*Monthly trends*,
*Filtering*,
*Time analysis*

```python
date_columns=[
    'order_purchase_timestamp',
    'order_approved_at',
    'order_delivered_carrier_date',
    'order_delivered_customer_date',
    'order_estimated_delivery_date'
]
for col in date_columns:
    orders[col]=pd.to_datetime(orders[col])
```

```python
orders.info()
```

Replacing the missing Values in **product_category_name** with **Unknown**

Missing cause problem in 
*Grouping*,
*Charts*,
*Revenue analysis*

```python
products['product_category_name'] = products[
    'product_category_name'
].fillna('Unknown')
```

Merging

Merge **order_items + orders**

order_items contain **revenue**

orders contain **timestamp** and **status**

*Sales Trend Analysis*


```python
sales_data = pd.merge(
    order_items,
    orders,
    on='order_id',
    how='left'
)
```

```python
sales_data.head()
```

```python
sales_data.shape
```

Merging **sales_data + customers**

customers contain **state** and **city**

*Regional Analysis*

```python
sales_data = pd.merge(
    sales_data,
    customers,
    on='customer_id',
    how='left'
)
```

```python
sales_data.head()
```

```python
sales_data.shape
```

Merging **sales_data** + **products**

products contain **products_categories** and **product_attributes**

```python
sales_data = pd.merge(
    sales_data,
    products,
    on='product_id',
    how='left'
)
```

```python
sales_data.columns
```

Merging **sales_data** + **category_name_translation**

**WHY ?**

*left_on* & *right_on*

Pandas renamed the column name while merging the Sales_data with product, because of the same column name already existed. 

```python
sales_data = pd.merge(
    sales_data,
    category_name_translation,
    left_on='product_category_name_x',
    right_on='product_category_name',
    how='left'
)
```

```python
sales_data.columns
```

Renaming **product_category_name_english** to **category_english**

Improve Readability

```python
sales_data.rename(
    columns={
        'product_category_name_english': 'category_english'
    },
    inplace=True
)
```

Creating **total_revenue*

```python
sales_data['total_revenue']=(
    sales_data['price']+
    sales_data['freight_value']
)
```
